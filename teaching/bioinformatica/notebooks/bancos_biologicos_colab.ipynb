## 5. Ler o FASTA com Python


```python
arquivo = "/content/drive/MyDrive/Bioinformatica_Biologia_Molecular/01_bancos/MK270576.1.fasta"

with open(arquivo) as f:
    linhas = f.readlines()

cabecalho = linhas[0].strip()
sequencia = "".join(linha.strip() for linha in linhas[1:])

print(cabecalho)
print(sequencia[:100])
```

## 6. Comprimento da sequência


```python
print(f"Comprimento: {len(sequencia)} bp")
```

## 7. Conteúdo GC

Conteúdo GC = `(G + C) / comprimento total × 100`.


```python
gc = (sequencia.count("G") + sequencia.count("C")) / len(sequencia) * 100

print(f"Comprimento: {len(sequencia)} bp")
print(f"GC: {gc:.2f}%")
```
